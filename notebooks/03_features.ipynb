## 1. Cechy sezonowe

```python
print("Kolumny sezonowe:")
print(final[['month', 'sezon', 'doy', 'sin_doy', 'cos_doy']].head(10))
```

#### Opis stworzonych cech sezonowych

- **month** — numer miesiąca (1-12), do mapowanie pory roku
- **sezon** — pora roku (zima/wiosna/lato/jesien), pozwala analizować 
  zależności osobno dla każdej pory roku
- **doy** — numer dnia roku (1-365)
- **sin_doy / cos_doy** — cykliczna reprezentacja dnia roku. 
  Zwykły numer miesiąca sugeruje że styczeń(1) i grudzień(12) są daleko 
  od siebie, a tak naprawde obok siebie. Sin/cos pokazuje tę cykliczność.

## 2. Cechy ciśnienia

```python
print("Kolumny ciśnienia:")
print(final[['Ciśnienie_średnia', 'Ciśnienie_min', 'Ciśnienie_max','Ciśnienie_ampl','Ciśnienie_delta_1d','Ciśnienie_delta_2d','Ciśnienie_delta_3d','Ciśnienie_trend_3d', 'Ciśnienie_trend_7d']].head(10))
```

#### Opis stworzonych cech ciśnienia

- **Ciśnienie_średnia / min / max** — dobowe statystyki ciśnienia
- **Ciśnienie_ampl** — amplituda ciśnienia w ciągu dnia (max - min)
- **Ciśnienie_delta_1d / 2d / 3d** — zmiana ciśnienia względem 1/2/3 dni wcześniej
- **Ciśnienie_trend_3d / 7d** — średnia ciśnienia z 3 i 7 dni

## 3. Proxy wiatru

```python
print("Proxy wiatru:")
print(final[['Wiatr_kierunek_proxy', 'Wiatr_sektor_proxy', 
             'Wiatr_sin_proxy', 'Wiatr_cos_proxy']].head(10))
print("\nRozkład sektorów:")
print(final['Wiatr_sektor_proxy'].value_counts())
```

#### Opis stworzonych kolumn z proxy

**Nie sa to prawdziwe dane o wietrze, zostały one wygenerowane na podstawie amplitudy cisnienia**
- **Wiatr_siła_proxy** — bezwzględna zmiana ciśnienia
- **Wiatr_kierunek_proxy** — znak zmiany (-1/0/1)
- **Wiatr_sektor_proxy** — kategoria: spadek/stabilnie/wzrost
- **Wiatr_kąt_proxy / sin / cos** — reprezentacja kierunku w postaci liczby
